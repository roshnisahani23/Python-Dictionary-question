Homework:-Find Topper in pysics Subject,math Subject ,Chemistry Subject.

```python
students = {
    1: {"Physics": 89, "Maths": 90, "Chemistry": 80},
    2: {"Physics": 70, "Maths": 98, "Chemistry": 92},
    3: {"Physics": 78, "Maths": 96, "Chemistry": 95}
}

subjects = ["Physics", "Maths", "Chemistry"]

for subject in subjects:
    topper = 0
    highest_marks = 0

    for roll_no in students:
        marks = students[roll_no][subject]

        if marks > highest_marks:
            highest_marks = marks
            topper = roll_no

    print(subject, "topper is Roll number", topper,
          "with marks", highest_marks)
```
